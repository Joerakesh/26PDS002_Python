```python
# Stone Paper Scissor
# ex 1 (Conditional flow statements)
# Stone Paper Scissor using Menu
import random

options = ["Stone", "Paper", "Scissor"]

point_human = 0
point_comp = 0

rounds = int(input("Enter how many rounds you want to play: "))

for i in range(rounds):
    print(f"\n========== Round {i+1} ==========")
    print("Choose your option:")
    print("1. Stone")
    print("2. Paper")
    print("3. Scissor")

    choice = int(input("Enter your choice (1-3): "))

    # Validate input
    if choice not in [1, 2, 3]:
        print("Invalid choice! Please select 1, 2, or 3.")
        continue

    human = options[choice - 1]
    comp = random.choice(options)

    print("Your choice     :", human)
    print("Computer choice :", comp)

    if human == comp:
        print("It's a Tie!")
        point_human += 1
        point_comp += 1

    elif (
        (human == "Stone" and comp == "Scissor") or
        (human == "Paper" and comp == "Stone") or
        (human == "Scissor" and comp == "Paper")
    ):
        print("You Win this round!")
        point_human += 1

    else:
        print("Computer Wins this round!")
        point_comp += 1

print("\n========== Final Result ==========")
print("Your Score     :", point_human)
print("Computer Score :", point_comp)

if point_human > point_comp:
    print("Congratulations! You Won the Game.")
elif point_comp > point_human:
    print("Computer Won the Game.")
else:
    print("The Game is a Tie.")
```

```python
# Ex 2:lambda function

from functools import reduce
transactions = [
    {"id":1,"amount_in_eur":700,"status":"Completed"},
    {"id":2,"amount_in_eur":650,"status":"returned"},
    {"id":3,"amount_in_eur":900,"status":"canceled"},
    {"id":4,"amount_in_eur":500,"status":"Completed"},
]

inr = 109.02

completed_tsnx = list(filter(lambda tsx:tsx["status"]=="Completed" , transactions))

print("Completed transactions")
print(completed_tsnx)

completed_eur_to_inr = list(map(lambda tsx:tsx["amount_in_eur"] * inr , completed_tsnx))

print("transactions completed in inr")
print(completed_eur_to_inr)

total_completed_inr = reduce(lambda x, y: x + y, completed_eur_to_inr)
print("\nTotal amount of completed transactions in INR:")
print(total_completed_inr)
```

```python
#ex 3 (Compose Functions)
import html
import functools

def compose(*functions):
    return lambda x: functools.reduce(lambda v, f: f(v), reversed(functions), x)

def remove_spaces(text: str) -> str:
    print("-> Running remove_spaces")
    return text.replace("p", "1")

def trim_spaces(text: str) -> str:
    print("-> Running trim_spaces")
    print(text.strip())
    return text.strip()

def escape_html_tags(text: str) -> str:
    print("-> Running escape_html_tags")
    print(html.escape(text))
    return html.escape(text)

sanitize_username = compose(
    remove_spaces,
    trim_spaces,
    escape_html_tags
)

sanitize_username(" <password'> ")
```

```python
from abc import ABC, abstractmethod


# ---------------- ABSTRACT CLASS ----------------
class Payment(ABC):

    def __init__(self, amount):
        self.amount = amount

    @abstractmethod
    def pay(self):
        pass

    @abstractmethod
    def display_receipt(self):
        pass


# ---------------- CASH PAYMENT ----------------
class Cash(Payment):

    def __init__(self, amount):
        super().__init__(amount)
        self.cash_received = 0
        self.balance = 0

    def pay(self):
        print("\nAvailable Notes:")
        print("₹500  ₹200  ₹100  ₹50  ₹20  ₹10")

        notes = [500, 200, 100, 50, 20, 10]

        total = 0

        for note in notes:
            count = int(input(f"How many ₹{note} notes? : "))
            total += note * count

        self.cash_received = total
        self.balance = total - self.amount

    def display_receipt(self):
        print("\n========== RECEIPT ==========")
        print(f"Payment Method : Cash")
        print(f"Bill Amount    : ₹{self.amount}")
        print(f"Cash Given     : ₹{self.cash_received}")

        if self.balance < 0:
            print(f"Still Need     : ₹{-self.balance}")
            print("Payment Status : Incomplete")
        else:
            print(f"Balance Return : ₹{self.balance}")
            print("Payment Status : Successful")

        print("=============================\n")


# ---------------- UPI PAYMENT ----------------
class UPI(Payment):

    def __init__(self, amount):
        super().__init__(amount)
        self.upi_id = ""

    def pay(self):
        self.upi_id = input("Enter your UPI ID: ")

        print("\nProcessing Payment...")
        print("Connecting to Bank...")
        print("Payment Successful ✔")

    def display_receipt(self):
        print("\n========== RECEIPT ==========")
        print(f"Payment Method : UPI")
        print(f"Bill Amount    : ₹{self.amount}")
        print(f"UPI ID         : {self.upi_id}")
        print("Payment Status : Successful")
        print("=============================\n")


# ---------------- MAIN PROGRAM ----------------
def main():

    print("=" * 40)
    print("      PAYMENT SYSTEM")
    print("=" * 40)

    amount = float(input("Enter Bill Amount: ₹"))

    print("\nChoose Payment Method")
    print("1. Cash")
    print("2. UPI")

    choice = input("Enter choice: ")

    if choice == "1":
        payment = Cash(amount)

    elif choice == "2":
        payment = UPI(amount)

    else:
        print("Invalid Choice")
        return

    payment.pay()
    payment.display_receipt()


if __name__ == "__main__":
    main()
```

```python
# Ex: 6 - Higher Order Function

# Outer function
def multiplication(multiplier):

    # Inner function
    def multiply(number):
        return number * multiplier

    # Return the inner function
    return multiply


# Create a function that always multiplies by 2
multiply_by_2 = multiplication(2)

# Call the returned function
result = multiply_by_2(4)

print("Result:", result)
```

```python
# Ex 7: Train and test model using oops
#!/usr/bin/env python
# coding: utf-8

# =====================================
# IMPORT LIBRARIES
# =====================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.linear_model import Ridge
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

# =====================================
# LOAD DATASET
# =====================================

df = pd.read_csv("Housing.csv")

print("=" * 60)
print("FIRST 5 ROWS")
print("=" * 60)
print(df.head())

# =====================================
# BASIC DATA EXPLORATION
# =====================================

print("\nDataset Shape :", df.shape)

print("\nDataset Information")
print(df.info())

print("\nStatistical Summary")
print(df.describe())

# =====================================
# CHECK MISSING VALUES
# =====================================

print("\nMissing Values")
print(df.isnull().sum())

# Fill numerical missing values (if any)
numeric_cols = df.select_dtypes(include=np.number).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

# =====================================
# REMOVE DUPLICATES
# =====================================

print("\nDuplicate Rows :", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Shape After Removing Duplicates :", df.shape)

# =====================================
# ONE-HOT ENCODING
# =====================================

df = pd.get_dummies(df, drop_first=True)

# =====================================
# SPLIT FEATURES AND TARGET
# =====================================

X = df.drop("price", axis=1)
y = df["price"]

# =====================================
# TRAIN TEST SPLIT
# =====================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

# =====================================
# FEATURE SCALING
# =====================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# =====================================
# POLYNOMIAL FEATURES
# =====================================

poly = PolynomialFeatures(
    degree=2,
    include_bias=False
)

X_train_poly = poly.fit_transform(X_train_scaled)
X_test_poly = poly.transform(X_test_scaled)

print("\nOriginal Features :", X_train.shape[1])
print("Polynomial Features :", X_train_poly.shape[1])

# =====================================
# RIDGE REGRESSION MODEL
# =====================================

model = Ridge(alpha=1.0)

model.fit(X_train_poly, y_train)

# =====================================
# PREDICTIONS
# =====================================

train_pred = model.predict(X_train_poly)
test_pred = model.predict(X_test_poly)

# =====================================
# MODEL EVALUATION
# =====================================

train_r2 = r2_score(y_train, train_pred)
test_r2 = r2_score(y_test, test_pred)

mae = mean_absolute_error(y_test, test_pred)
mse = mean_squared_error(y_test, test_pred)
rmse = np.sqrt(mse)

print("\n" + "=" * 60)
print("MODEL PERFORMANCE")
print("=" * 60)

print(f"Training R² : {train_r2:.4f}")
print(f"Testing R²  : {test_r2:.4f}")

print(f"\nMAE  : {mae:.2f}")
print(f"MSE  : {mse:.2f}")
print(f"RMSE : {rmse:.2f}")

# =====================================
# CROSS VALIDATION
# =====================================

cv_scores = cross_val_score(
    Ridge(alpha=1.0),
    X_train_poly,
    y_train,
    cv=5,
    scoring="r2"
)

print("\nCross Validation Scores")
print(cv_scores)

print("Average CV Score :", cv_scores.mean())

# =====================================
# OVERFITTING CHECK
# =====================================

print("\n" + "=" * 60)

difference = abs(train_r2 - test_r2)

if difference < 0.05:
    print("Model is well generalized.")
elif difference < 0.15:
    print("Model has slight overfitting.")
else:
    print("Model is overfitting.")

print("=" * 60)

# =====================================
# ACTUAL VS PREDICTED
# =====================================

plt.figure(figsize=(8,6))

plt.scatter(y_test, test_pred)

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linewidth=2
)

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual vs Predicted House Prices")

plt.show()

# =====================================
# SAMPLE PREDICTIONS
# =====================================

comparison = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": test_pred
})

print("\nFirst 10 Predictions")
print(comparison.head(10))
```
