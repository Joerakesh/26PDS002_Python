```python
# ex 1 (Conditional flow statements)
# Stone Paper Scissor using Menu
import random

options = ["Stone", "Paper", "Scissor"]

point_human = 0
point_comp = 0

rounds = int(input("Enter how many rounds you want to play: "))

for i in range(rounds):
    print(f"\n========== Round {i+1} ==========")
    print("Choose your option:")
    print("1. Stone")
    print("2. Paper")
    print("3. Scissor")

    choice = int(input("Enter your choice (1-3): "))

    # Validate input
    if choice not in [1, 2, 3]:
        print("Invalid choice! Please select 1, 2, or 3.")
        continue

    human = options[choice - 1]
    comp = random.choice(options)

    print("Your choice     :", human)
    print("Computer choice :", comp)

    if human == comp:
        print("It's a Tie!")
        point_human += 1
        point_comp += 1

    elif (
        (human == "Stone" and comp == "Scissor") or
        (human == "Paper" and comp == "Stone") or
        (human == "Scissor" and comp == "Paper")
    ):
        print("You Win this round!")
        point_human += 1

    else:
        print("Computer Wins this round!")
        point_comp += 1

print("\n========== Final Result ==========")
print("Your Score     :", point_human)
print("Computer Score :", point_comp)

if point_human > point_comp:
    print("Congratulations! You Won the Game.")
elif point_comp > point_human:
    print("Computer Won the Game.")
else:
    print("The Game is a Tie.")
```
